```python
import numpy as np

def gen_df(rows=2):
    cols = ["A", "B", "C", "D"]
    binary = np.random.binomial(n=1, p=0.5, size=(rows, 1))
    data = np.random.normal(loc=0, scale=5, size=(rows,2))
    uni = np.random.uniform(low=-10, high=10, size=(rows,1))
    return pd.DataFrame(np.concat((binary,data, uni), axis=1), columns=cols)

dfs = {f"label{i}" : gen_df() for i in range(5)}

df = pd.concat(dfs, names=["index 0","tomato 1"])
print(df)
df = df.reset_index(level=1)
print(df)
df = df.reset_index(drop=True)
print(df)
```

```python
import numpy as np
mu, sigma = 0, 0.000000001
x = np.arange(10).reshape(-1,1).repeat(10, axis=1)
y = x.transpose()

print(x)
print(y.transpose())
print(y)

x = x + np.random.normal(loc=mu,scale=sigma, size=x.shape)
y = y + np.random.normal(loc=mu,scale=sigma, size=y.shape)
print(np.allclose(np.diff(x), 0))
print(np.allclose(np.diff(y), 0))
```
